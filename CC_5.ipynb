## Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations:
### a) Addition of 2 in all the elements
### b) Multiply 3 with all the elements
### c) Divide every element by 2


```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print("Original Array:", arr)

print("\na) Addition of 2 in all elements:")
print(arr + 2)

print("\nb) Multiply 3 with all elements:")
print(arr * 3)

print("\nc) Divide every element by 2:")
print(arr / 2)
```

## Q.2 Questions on Basic NumPy Array:
### a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])
### b) Find the most frequent value and their indice(s) in:
- i. x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
- ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])


```python
# a) Reverse
arr_rev = np.array([1, 2, 3, 6, 4, 5])
print("Original Array:", arr_rev)
print("a) Reversed Array:", arr_rev[::-1])

# b) i. Most frequent in x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
vals_x, counts_x = np.unique(x, return_counts=True)
mode_x = vals_x[np.argmax(counts_x)]
indices_x = np.where(x == mode_x)[0]
print(f"\nb) i. Array x: {x}")
print(f"   Most frequent value: {mode_x}, Indices: {indices_x}")

# b) ii. Most frequent in y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
vals_y, counts_y = np.unique(y, return_counts=True)
max_c_y = np.max(counts_y)
modes_y = vals_y[counts_y == max_c_y]
print(f"\nb) ii. Array y: {y}")
for m in modes_y:
    indices_y = np.where(y == m)[0]
    print(f"   Most frequent value: {m}, Indices: {indices_y}")
```

## Q.4 Create 1-D NumPy array named `Student_1337` with 25 evenly spaced numbers from 10 to 100 using linspace(). Print dimensions, shape, total elements, dtype, and total bytes consumed.


```python
Student_1337 = np.linspace(10, 100, 25)
print("Array Student_1337:\n", Student_1337)

print("\nArray Attributes:")
print("  Dimensions (ndim):", Student_1337.ndim)
print("  Shape:", Student_1337.shape)
print("  Total elements (size):", Student_1337.size)
print("  Data type:", Student_1337.dtype)
print("  Total bytes consumed (nbytes):", Student_1337.nbytes)
```

## Q.5 Create 2-D Array (3 rows, 4 columns) named `ucs420_1337` with values: 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute mean, median, max, min, unique elements. Reshape to 4 rows and 3 columns, compute row-wise and column-wise sum.


```python
ucs420_1337 = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print("Original 2D Array ucs420_1337 (3x4):\n", ucs420_1337)

print("\nStatistical Reductions:")
print("  Mean:", np.mean(ucs420_1337))
print("  Median:", np.median(ucs420_1337))
print("  Maximum:", np.max(ucs420_1337))
print("  Minimum:", np.min(ucs420_1337))
print("  Unique elements:", np.unique(ucs420_1337))

ucs420_1337_reshaped = ucs420_1337.reshape(4, 3)
print("\nReshaped Array (4x3):\n", ucs420_1337_reshaped)

print("\nRow-wise sum (axis=1):", np.sum(ucs420_1337_reshaped, axis=1))
print("Column-wise sum (axis=0):", np.sum(ucs420_1337_reshaped, axis=0))
print("Total sum:", np.sum(ucs420_1337_reshaped))
```
