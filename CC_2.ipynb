### Q1 (i). Take roll number (1024171337). Extract individual digits and multiply each by 10 to form list L.


```python
roll = 1024171337
digits = [int(d) for d in str(roll)]
L = [d * 10 for d in digits]
print("Roll number:", roll)
print("Extracted digits:", digits)
print("List L (digits * 10):", L)
```

### Q1 (ii). Add two numbers to L — one using append() and one using insert() at a specific position. Explain in a comment what changed.


```python
L.append(55)
print("After L.append(55):", L)
# append() adds the element 55 at the end of list L

L.insert(3, 77)
print("After L.insert(3, 77):", L)
# insert(3, 77) places 77 at index 3, right-shifting existing elements
```

### Q1 (iii). Remove two elements from L — one using remove() and one using pop().


```python
L.remove(55)
print("After L.remove(55):", L)
# remove() searches for the first occurrence of the value 55 and deletes it

popped_val = L.pop()
print(f"Popped value: {popped_val}")
print("After L.pop():", L)
# pop() removes and returns the last element from the list
```

### Q3. Random Numbers with Seed: Generate 100 random numbers between 100 and 900.


```python
import random

random.seed(1024171337)
rand_list = [random.randint(100, 900) for _ in range(100)]
print("Total random numbers generated:", len(rand_list))
print("Sample first 15 numbers:", rand_list[:15])
```

### Q3 (ii & iii). Count and print odd and even numbers in the list.


```python
odd_nums = [x for x in rand_list if x % 2 != 0]
even_nums = [x for x in rand_list if x % 2 == 0]

print(f"Total Odd Numbers: {len(odd_nums)}, Sample: {odd_nums[:8]}")
print(f"Total Even Numbers: {len(even_nums)}, Sample: {even_nums[:8]}")
```

### Q3 (iv & v). Prime numbers detection and most frequent value.


```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_nums = [x for x in rand_list if is_prime(x)]
print(f"Total Primes Found: {len(prime_nums)}")
print("Prime Numbers:", prime_nums)

most_freq = max(rand_list, key=rand_list.count)
print(f"\nMost frequent number: {most_freq}, Frequency: {rand_list.count(most_freq)}")
```

### Q4. Set Operations: A = {digit * 7} and B = {digit * 9}.


```python
raw_digits = digits[:8]
A = {d * 7 for d in raw_digits}
B = {d * 9 for d in raw_digits}

print("Set A:", A)
print("Set B:", B)

print("\nUnion (A ∪ B):", A.union(B))
print("Intersection (A ∩ B):", A.intersection(B))
print("Difference A - B:", A.difference(B))
print("Difference B - A:", B.difference(A))
# difference() gives values unique to one set, whereas symmetric_difference() gives values in either set but not both.

print("Symmetric Difference (A Δ B):", A.symmetric_difference(B))
print("A is subset of B?:", A.issubset(B))
print("B is superset of A?:", B.issuperset(A))

val_x = 70
A.discard(val_x)
print(f"After A.discard({val_x}):", A)
# discard() removes the item if present without raising an error if absent, unlike remove().
```

### Q5. Dictionary Operations: Personalized dictionary and operations.


```python
my_dict = {
    "name": "meet-the-1337",
    "roll_no": "1024171337",
    "branch": "COPC",
    "age": 20,
    "city": "Patiala"
}
print("Original dictionary:", my_dict)

# i. Rename 'city' to 'location'
my_dict["location"] = my_dict.pop("city")
print("After renaming 'city' -> 'location':", my_dict)

# ii. Add 'cgpa'
my_dict["cgpa"] = 8.6
print("After adding 'cgpa':", my_dict)

# iii. Increment 'age' by 1
my_dict["age"] += 1
print("After incrementing 'age':", my_dict)

# iv. Delete 'branch' using pop() and del
d1 = my_dict.copy()
d2 = my_dict.copy()
popped_branch = d1.pop("branch")
del d2["branch"]
print(f"pop() returned '{popped_branch}', remaining d1: {d1}")
print(f"del d2['branch'] result: {d2}")
# pop() deletes and returns the associated value; del removes the key in-place and returns nothing.

# v. Iterate with .items()
print("\nKey-Value Pairs:")
for k, v in my_dict.items():
    print(f"  {k} -> {v}")

# vi. Safe check for 'email'
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Safe fallback: 'email' key not found.")

# vii. Merge with friend_dict
friend_dict = {
    "name": "Alex",
    "roll_no": "1024171338",
    "branch": "COPC",
    "age": 21,
    "city": "Mohali"
}
merged_dict = {**my_dict, **friend_dict}
print("\nMerged dictionary:", merged_dict)
# When keys collide, values from the rightmost dictionary overwrite earlier entries.

# viii. Dictionary comprehension for string values
str_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("String entries only:", str_dict)
```
