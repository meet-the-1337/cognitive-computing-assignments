Q1. Consider the following sensor readings:
`temperature = np.array([24, 27, 30, 34, 37, 26, 32, 39])`
A sensor has an average measurement error of +2°C.
Perform the following without using a for loop:
a. Add 2°C to every reading using vectorization.
b. Convert the corrected temperatures to Fahrenheit: F = (9/5) * C + 32.
c. Use boolean indexing to extract all temperature readings strictly greater than 32°C.
d. Count how many temperature readings exceed 32°C.
e. In a comment, explain how vectorization and boolean indexing contribute to efficient numerical data processing.


````python
import numpy as np

temperature = np.array([24, 27, 30, 34, 37, 26, 32, 39])

temperature_corrected = temperature + 2
print("a. Corrected temperatures (+2°C):", temperature_corrected)

fahrenheit = (9 / 5) * temperature_corrected + 32
print("b. Temperatures in Fahrenheit:", fahrenheit)

greater_than_32 = temperature_corrected[temperature_corrected > 32]
print("c. Readings strictly > 32°C:", greater_than_32)

count_32 = np.sum(temperature_corrected > 32)
print("d. Count of readings > 32°C:", count_32)
````

**Answer:**
- **Vectorization** applies arithmetic and mathematical operations across entire arrays at the compiled C-level SIMD level, eliminating Python bytecode iteration overhead and accelerating processing.
- **Boolean Indexing** produces boolean bitmasks evaluated elementwise in memory, allowing instantaneous filtering and conditional selection without nested conditionals.


Q2. The following matrix represents the number of daily steps recorded by four users over three days:
```python
steps = np.array([
    [5200, 6300, 7200],
    [8100, 7600, 9100],
    [4600, 5200, 4900],
    [9100, 8600, 9600]
])
```
Perform the following:
a. Calculate the total steps taken across all users and all days.
b. Calculate the average (mean) steps taken per entry.
c. Find the maximum and minimum number of steps in the entire array.
d. Calculate the total steps taken on each day (column-wise sum).
e. Calculate the total steps taken by each user (row-wise sum).
f. Identify which user and on which day the highest number of steps was recorded using `np.argmax()` and `np.unravel_index()`.


````python
steps = np.array([
    [5200, 6300, 7200],
    [8100, 7600, 9100],
    [4600, 5200, 4900],
    [9100, 8600, 9600]
])

print("a. Total steps across all users:", np.sum(steps))
print("b. Mean steps per entry:", np.mean(steps))
print(f"c. Maximum steps: {np.max(steps)}, Minimum steps: {np.min(steps)}")

daily_totals = np.sum(steps, axis=0)
print("d. Total steps for each day (cols):", daily_totals)

user_totals = np.sum(steps, axis=1)
print("e. Total steps for each user (rows):", user_totals)

max_pos = np.unravel_index(np.argmax(steps), steps.shape)
print(f"f. Peak activity: User {max_pos[0] + 1} on Day {max_pos[1] + 1} with {steps[max_pos]} steps (Index: {max_pos})")
````

Q3. Write a Python program using NumPy to perform the following operations:
a. Create `original = np.array([2, 4, 6, 8, 10, 12])`.
b. Create a slice containing elements from index 1 to 4 (`subset`).
c. Modify the first element of `subset` to 888. Print both `original` and `subset`. Explain the result.
d. Repeat the previous step, but create an independent copy using `.copy()` and modify its first element to 400. Print both.
e. Reshape `np.arange(1, 13)` into a 3x4 matrix. Access rows, columns, submatrices.
f. Flatten the matrix using `flatten()` and `ravel()`. Modify elements in each and observe effects on original.


````python
original = np.array([2, 4, 6, 8, 10, 12])
subset = original[1:5]
subset[0] = 888
print("After modifying slice subset[0] = 888:")
print("  Original:", original)
print("  Subset:  ", subset)

copied_array = original[1:5].copy()
copied_array[0] = 400
print("\nAfter modifying copied_array[0] = 400:")
print("  Original:    ", original)
print("  Copied array:", copied_array)

matrix = np.arange(1, 13).reshape(3, 4)
print("\n3x4 Matrix:\n", matrix)
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Submatrix (rows 1-2, cols 2-3):\n", matrix[1:3, 2:4])

raveled = matrix.ravel()
raveled[0] = 1000
print("\nAfter mutating raveled[0] = 1000:")
print("  Matrix:\n", matrix)

flattened = matrix.flatten()
flattened[1] = 2000
print("\nAfter mutating flattened[1] = 2000:")
print("  Matrix:\n", matrix)
````

**Answer:**
- Standard NumPy slicing returns a **view** sharing the underlying data buffer; modifying the slice mutates the original array.
- `.copy()` allocates new, separate memory, ensuring changes do not impact the source.
- `ravel()` returns a flattened view whenever possible (mutating original), while `flatten()` always returns a deep contiguous copy.


Q4. Cognitive-Assistive System Linear Model:
Predict a user's assistance score based on:
- Sleep hours
- Activity level
- Stress level
Consider:
```python
X = np.array([
    [7, 65, 4],
    [6, 55, 5],
    [9, 75, 3],
    [5, 35, 7]
])
y = np.array([42, 63, 35, 82])
```
Compute transpose $X^T$, Gram matrix $X^T X$, inverse $(X^T X)^{-1}$, normal equation weights $eta = (X^T X)^{-1} X^T y$, and predict assistance score for new user profile `[6, 45, 6]`.


````python
X = np.array([
    [7, 65, 4],
    [6, 55, 5],
    [9, 75, 3],
    [5, 35, 7]
])
y = np.array([42, 63, 35, 82])

print("Feature Matrix X (4x3):\n", X)
print("Target y:", y)

XT = X.T
print("\nX Transpose (X.T):\n", XT)

XTX = XT @ X
print("\nGram Matrix (X.T @ X):\n", XTX)

XTX_inv = np.linalg.inv(XTX)
print("\nInverse of (X.T @ X):\n", XTX_inv)

beta = XTX_inv @ XT @ y
print("\nEstimated Regression Coefficients (beta):", beta)

new_user = np.array([6, 45, 6])
predicted_score = new_user @ beta
print(f"\nPredicted Assistance Score for Profile [6, 45, 6]: {predicted_score:.2f}")
````

**Answer:**
- The matrix transpose $X^T$ flips rows into columns, transforming dimension $(4 \times 3)$ to $(3 \times 4)$ to enable inner product matrix multiplication.
- The coefficients $\beta$ capture the partial derivative weight of each physical feature (sleep, activity, stress) with respect to the user assistance requirement.
