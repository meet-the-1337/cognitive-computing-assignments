## Part 3: Loops (While, For, Factorial)


```python
print("WHILE LOOP (Odd squares up to 10):")
i = 1
while i <= 10:
    print(f"  {i}^2 = {i * i}")
    i += 2

print("\nFOR LOOP (Cubes in range(0, 10, 2)):")
for i in range(0, 10, 2):
    print(f"  {i}^3 = {i ** 3}")

print("\nFACTORIAL COMPUTATION:")
num = 10
fact = 1
for x in range(1, num):
    fact *= x
print(f"  Factorial of {num - 1}: {fact}")
```

## Part 4: Conditionals & Prime Number Verification


```python
def check_even_odd(n):
    return "even" if n % 2 == 0 else "odd"

print("1337 is:", check_even_odd(1337))

def isPrime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

test_numbers = [11, 15, 23, 29, 35, 47]
print("\nPrime Number Checks:")
for val in test_numbers:
    print(f"  {val} -> isPrime: {isPrime(val)}")
```
